```python
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
```

```python
cities_clean = []
for i in cities:
    city = i.strip().capitalize()
    cities_clean.append(city)
print(cities_clean)
```

```python
amounts_int = []
for i in amounts:
    amount = int(i)
    amounts_int.append(amount)
print(amounts_int)
```

```python
count_orders = len(amounts_int)
general_sum = sum(amounts_int)
avg = round(general_sum/count_orders)
print(count_orders, general_sum, avg)
```

```python
unpaid_count = 0
unpaid_sum = 0
for i in range(len(paid)):
    if paid[i] == False:
        unpaid_count += 1
        unpaid_sum += amounts_int[i]
print(unpaid_count, unpaid_sum)
```

```python
for city, amount, status in zip(cities_clean, amounts_int, paid):
    print(f"{city} | {amount} | {status}")
```

```python
almaty_revenue = 0
astana_revenue = 0
for city, amount, status in zip(cities_clean, amounts_int, paid):
    if city == "Almaty" and status == True:
        almaty_revenue += amount
    if city == "Astana" and status == True:
        astana_revenue += amount
print(almaty_revenue, astana_revenue)
```

```python
max_amount = 0
max_index = -1
for i in range(len(amounts_int)):
    if paid[i] == True and amounts_int[i] > max_amount:
        max_amount = amounts_int[i]
        max_index = i
print(max_index, cities_clean[max_index], max_amount)
```

```python
weekend_count = 0
for i in range(len(days)):
    if days[i] == "сб" or days[i] == "вс":
        weekend_count += 1
weekend_percent = weekend_count / count_orders * 100
print(weekend_percent)
```
